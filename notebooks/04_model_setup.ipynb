This notebook verifies that the ML training environment is correctly set up.
It loads processed datasets and ensures they are ready for model training.

The dataset is highly imbalanced (~0.13% fraud cases), which will impact model training and evaluation.

```python
from pyspark.sql import SparkSession

spark = SparkSession.builder \
    .appName("ModelSetup") \
    .master("local[*]") \
    .getOrCreate()

train_df = spark.read.parquet("../data/processed/train")
test_df = spark.read.parquet("../data/processed/test")

# schema check
train_df.printSchema()

# sample
train_df.show(5)

# dataset size
print("Train:", train_df.count())
print("Test:", test_df.count())

# class distribution
train_df.groupBy("label").count().show()

print("ML dataset ready: features + label")
```
